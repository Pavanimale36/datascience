```python
arr1= np.array([1,2,3])
arr2= np.array([10,20,30])
print(arr1+arr2)
#subtration
#10-1=9
#20-2=18
#30-3=27
print(arr2-arr1)
#multiplication
#10*1=10
#20*2=20
#30*1=30
print(arr1*arr2)
# #division
#10/1=10
#20/2=10
#30/3=10
print(arr2/arr1)
print(arr2//arr1)
print(arr2%arr1)
print(arr1**2)
```

```python
#aggregate function:functions used to summarize data
```

```python
l=np.array([10,20,20,40,50,60])
#add all elements=sum
print(np.sum(l))
#average
print(np.mean(l))
#minimum
print(np.min(l))
#maximum value
print(np.max(l))
#standard devation
print(np.std(l))
#variance
print(np.var(l))
#size of element in an array
print(np.size(l))
#argmax:gives the index position of the largest value
print(np.argmax(l))
#argmin #it gives the index position of the minimum value
print(np.argmin(l))
#applying square root
print(np.sqrt(l))
```

```python
arr=np.array([-1,-2,-3,-4,-5])
print(np.abs(arr)) #absolute function converts negative to positive
```

```python
#numpy rounding functions
arr=np.array([1.2,2.7,3.5,4.1,5.9])
#round:round values to nearest integer
# 1.2->1
# 2.7->3
# 3.5->3
# 4.1->4
#5.9->6
print(np.round(arr))
```

```python
arr=np.array([1.2,2.7,3.5,4.1,5.9])
#floor it will always gives lower integer values
#1.2->1
#2.7->2
#3.5->3
#4.1->4
#5.9->5
print(np.floor(arr))
```
