```python
import numpy as np
```

```python
#addition
mat1=np.array([
    [1,2],
    [3,4]
])
mat2=np.array([
    [5,6],
    [7,8]
])
```

```python
result = mat1 + mat2
print(result)
```

```python
result = mat1 -mat2
print(result)
```

```python
result = mat1*mat2
print(result)
```

```python
result=np.matmul(mat1,mat2)
print(result)
result=mat1 @mat2
print(result)
result=np.dot(mat1,mat2)
print(result)
```

```python
result=mat1/mat2
print(result)
```

```python
mat=np.array([
    [1,2,3],
    [4,5,6]
])
print(mat.T)
```

```python
#matrix shape
print(mat.shape)
```

```python
#identity matrix
identity=np.eye(3)
print(identity)
identity=np.eye(2)
print(identity)
```

```python
#determinant
mat=np.array([
    [1,2],
    [3,4]
])
print(np.linalg.det(mat))
```

```python
#matrix inverse
mat=np.array([
    [1,2],
    [3,4]
])
print(np.linalg.inv(mat))
```
