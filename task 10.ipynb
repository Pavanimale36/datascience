```python
import pandas as pd
from datetime import datetime
df=pd.read_csv("/content/netflix_titles (3).csv")
df
```

```python
df["director"].fillna("unknown",inplace=True)
df
```

```python
#cast Missing Values
df["cast"].fillna("unknown",inplace=True)
df
```

```python
print(df.columns)
```

```python
df
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Not Rated')
df['description'] = df['description'].fillna('No Description')
df['date_added'] = df['date_added'].fillna('01-01-2000')
df
```

```python
# Convert date_added to datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df
```

```python
# Feature engineering
df['AddedYear'] = df['date_added'].dt.year
df
```

```python
df['AddedMonth'] = df['date_added'].dt.month
df
```

```python
df['AddedDay'] = df['date_added'].dt.day
df
```

```python
df['AddedWeekDay'] = df['date_added'].dt.day_name()
df
```

```python
df['AddedQuarter'] = df['date_added'].dt.quarter
df
```

```python
df['ContentAge'] = datetime.now().year - df['release_year']
df['YearsToNetflix'] = df['AddedYear'] - df['release_year']
df
```

```python
df['TitleLength'] = df['title'].apply(lambda x: len(str(x)))
df['TitleWordCount'] = df['title'].apply(lambda x: len(str(x).split()))
df
```

```python
df['LongTitle'] = df['TitleWordCount'].apply(lambda x: 1 if x > 3 else 0)
df
```

```python
df['DescriptionLength'] = df['description'].apply(lambda x: len(str(x)))
df['DescriptionWordCount'] = df['description'].apply(lambda x: len(str(x).split()))
df
```

```python
df['ActorCount'] = df['cast'].apply(lambda x: len(str(x).split(',')))
df['CountryCount'] = df['country'].apply(lambda x: len(str(x).split(',')))
df
```

```python
df['GenreCount'] = df['listed_in'].apply(lambda x: len(str(x).split(',')))
df
```

```python
df['MultiGenre'] = df['GenreCount'].apply(lambda x: 1 if x > 1 else 0)
df
```

```python
df['IsMovie'] = df['type'].apply(lambda x: 1 if x.lower() == 'movie' else 0)
df['IsTVShow'] = df['type'].apply(lambda x: 1 if x.lower() == 'tv show' else 0)
df
```

```python
df['RecentContent'] = df['release_year'].apply(lambda x: 1 if x >= 2020 else 0)
df['ClassicContent'] = df['release_year'].apply(lambda x: 1 if x < 2000 else 0)
df
```
