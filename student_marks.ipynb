```python
import pandas as pd
```

```python
student_df=pd.read_csv("/content/students.csv")
print(student_df)
```

```python
marks_df=pd.read_csv("/content/marks.csv")
print(marks_df)
```

```python
#inner merge
#it will take only common data from both tables
result=pd.merge(
  student_df ,
  marks_df ,
    on='student_id',
    how='inner'
)
print(result)
```

```python
#left merge
#allow rows from left dataframe
```

```python
result=pd.merge(
  student_df ,
  marks_df ,
    on='student_id',
    how='left'
)
print(result)
```

```python
result=pd.merge(
  student_df ,
  marks_df ,
    on='student_id',
    how='outer'
)
print(result)
```
