```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder

diamonds = sns.load_dataset('diamonds')

cut_column = diamonds[['cut']]

encoder = OneHotEncoder(sparse_output=False)

cut_encoded = encoder.fit_transform(cut_column)

cut_encoded_df = pd.DataFrame(
    cut_encoded,
    columns=encoder.get_feature_names_out(['cut'])
)

print(cut_encoded_df.head())
```

```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder

diamonds = sns.load_dataset('diamonds')

encoder = OneHotEncoder(sparse_output=False)

cut_encoded_df = pd.DataFrame(
    encoder.fit_transform(diamonds[['cut']]),
    columns=encoder.get_feature_names_out(['cut'])
)

color_encoded_df = pd.DataFrame(
    encoder.fit_transform(diamonds[['color']]),
    columns=encoder.get_feature_names_out(['color'])
)

clarity_encoded_df = pd.DataFrame(
    encoder.fit_transform(diamonds[['clarity']]),
    columns=encoder.get_feature_names_out(['clarity'])
)

diamonds_remaining = diamonds.drop(['cut', 'color', 'clarity'], axis=1)

final_df = pd.concat(
    [diamonds_remaining, cut_encoded_df, color_encoded_df, clarity_encoded_df],
    axis=1
)

print(final_df.head())
print(final_df.shape)
print(final_df.columns.tolist())
```
