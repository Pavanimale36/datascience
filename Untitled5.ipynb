```python
df.columns=(
    df.columns.str.lower().str.replace(" ","_")
)
df
```

```python
import pandas as pd
```

```python
df=pd.read_csv("/content/employee_data (1).csv")
df
```

```python
df["Age"].fillna(df["Age"].mean(),inplace=True);
df["Employee Name"].fillna("unknown",inplace=True)
df
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
df["Employee Name"]=df["Employee Name"].str.strip();df["Department"]=df["Department"].str.strip().str.title()
df
```

```python
df["Department"]=df["Department"].replace({'It':'information Technology'})
df
```

```python
df.rename(columns={'Employee Name':'employee_name','Salary':'salary','Join Date':'join date'},inplace=True)
df
```
