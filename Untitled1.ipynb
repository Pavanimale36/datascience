```python
#concat in pandas
concat() is used when you want to combine multiple dataframes
```

```python
import pandas as pd
india_weather=pd.DataFrame({
    "city":["mumbai","delhi","bangalore"],
    "temp":[30,35,40],
    "humidity":[70,55,60]
})
print(india_weather)
```

```python
paris_weather=pd.DataFrame(
    {
    "city":["tir","kadapa","dallaspuram"],
    "temp":[30,35,40],
    "humidity":[70,55,60]
    }
)
print(paris_weather)
```

```python
df=pd.concat([india_weather,paris_weather])
df
```

```python
df=pd.concat([india_weather,paris_weather],ignore_index=True)
df
```

```python
df=pd.concat([india_weather,paris_weather],keys=["ind","paris"])
df
```

```python
df=pd.concat([india_weather,paris_weather],axis=1)
df
```
